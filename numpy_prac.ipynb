```python
import requests
import os

# Ensure the TMDB_API_KEY environment variable is set
# Replace 'YOUR_ACTUAL_TMDB_API_KEY' with your valid TMDb API key
os.environ["TMDB_API_KEY"] = "YOUR_ACTUAL_TMDB_API_KEY"

tmdb_api_key = os.environ.get("TMDB_API_KEY")

if not tmdb_api_key or tmdb_api_key == "YOUR_ACTUAL_TMDB_API_KEY":
    print("ACTION REQUIRED: Please replace 'YOUR_ACTUAL_TMDB_API_KEY' with your valid TMDb API key in the code or environment variable.")
else:
    # Construct the URL to fetch the list of movie genres
    genre_url = f"https://api.themoviedb.org/3/genre/movie/list?api_key={tmdb_api_key}"
    try:
        # Make a GET request to the genre list URL
        genre_response = requests.get(genre_url)
        genre_response.raise_for_status() # Raise an exception for bad status codes
        genres = genre_response.json().get('genres', [])
        science_fiction_genre_id = None
        # Parse the JSON response to find the genre ID for 'Science Fiction'
        for genre in genres:
            if genre.get('name') == 'Science Fiction':
                science_fiction_genre_id = genre.get('id')
                break

        if science_fiction_genre_id:
            print(f"Science Fiction Genre ID: {science_fiction_genre_id}")

            # Construct the URL to discover movies with that genre ID
            discover_url = f"https://api.themoviedb.org/3/discover/movie?api_key={tmdb_api_key}&with_genres={science_fiction_genre_id}"
            # Make a GET request to the discover movies URL
            discover_response = requests.get(discover_url)
            discover_response.raise_for_status() # Raise an exception for bad status codes
            # Parse the JSON response and extract the movie data
            sci_fi_movies_data = discover_response.json().get('results', [])

            # Store the identified sci-fi movies in a list
            sci_fi_movies = []
            for movie in sci_fi_movies_data:
                sci_fi_movies.append({'id': movie.get('id'), 'title': movie.get('title')})

            # Print the number of found sci-fi movies and display the first 5
            print(f"Found {len(sci_fi_movies)} Science Fiction movies.")
            print("First 5 Science Fiction movies:")
            for movie in sci_fi_movies[:5]:
                print(movie)

        else:
            print("Science Fiction genre not found.")

    except requests.exceptions.RequestException as e:
        print(f"Error during API request: {e}")
    except Exception as e:
        print(f"An unexpected error occurred: {e}")
```
